```python
from mmseg.apis import MMSegInferencer
import cv2

img = cv2.imread("cam_00.jpg")
imgs = ['cam_00.jpg', 'demo.jpg']
# Resize để chia hết cho 32 (ví dụ: 1280x720)
resized = cv2.resize(img, (1280, 720))
inferencer = MMSegInferencer(model='deeplabv3plus_r101-d8_4xb2-80k_cityscapes-512x1024', device='cpu')

# Run inference and visualize
result = inferencer(imgs, show=True, opacity=0.5, out_dir='output/deeplabv3plus')
```
