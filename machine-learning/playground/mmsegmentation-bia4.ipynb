```python
from mmseg.apis import inference_model, init_model
import mmcv
import matplotlib.pyplot as plt

# Config and checkpoint paths
base_path = './'
config_file = base_path + 'pspnet_r50-d8_4xb2-40k_cityscapes-512x1024.py'
checkpoint_file = base_path + 'pspnet_r50-d8_512x1024_40k_cityscapes_20200605_003338-2966598c.pth'

# Initialize model with visualizer config
model = init_model(
    config_file,
    checkpoint_file,
    device='cpu',
    cfg_options={
        'visualizer': {
            'type': 'Visualizer',
            'vis_backends': [{'type': 'LocalVisBackend', 'save_dir':  base_path + 'output'}]
        }
    }
)

# Test a single image
img = base_path + 'cam_00.jpg'  # Ensure this path is valid
result = inference_model(model, img)

# Load image and create segmentation visualization
img_array = mmcv.imread(img)
vis_result = model.show_result(img_array, result, opacity=0.5)

# Display in notebook
plt.imshow(vis_result)
plt.axis('off')  # Hide axes
plt.show()
```
